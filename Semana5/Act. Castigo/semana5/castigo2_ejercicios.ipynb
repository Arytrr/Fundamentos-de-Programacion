```python
# Ejercico 1. Bitácora de la semana. 

dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def resumen(km):
    total = 0
    dias_rodados = 0
    mayor = 0
    indice = 0

    for i in range(len(km)):
        total += km[i]
        if km[i] > 0:
            dias_rodados += 1
        if km[i] > mayor:
            mayor = km[i]
            indice = i

    promedio = round(total / dias_rodados, 1)

    return total, dias_rodados, promedio, indice

total, dias_rodados, promedio, indice = resumen(km)

print("Total:", total, "km")
print("Días con rodada:", dias_rodados)
print("Promedio por día rodado:", promedio, "km")
print("Día más largo:", dias[indice], "(", km[indice], "km )")
```

```python
# Ejercicio 2. La votación del festival. 
votos = [("Metallica", 48), ("Slayer", 22), ("Megadeth", 35), ("Anthrax", 15), ("Iron Maiden", 40)]

def promedio_votos(lista): 
    total = 0

    for banda, voto in lista: 
        total += voto
    return total / len(lista)

mayor = 0 
ganador = ""

for banda, voto in votos: 
    if voto > mayor: 
        mayor = voto
        ganador = ganador

promedio = promedio_votos(votos)

print("Ganador:", ganador, "con", mayor, "votos")
print("Promedio:", promedio)
print("Arriba del promedio:")

for banda, voto in votos:
    if voto > promedio: 
        print("-", banda, "(", voto, ")")
```

```python
# Ejercicio 3. Tramos de la ruta a la sierra. 
paradas = (("Querétaro", 0), ("San Juan del Río", 52), ("Tequisquiapan", 74), ("Bernal", 125), ("Jalpan", 260))
mayor = 0
tramo = ""

for i in range(len(paradas) - 1): 
    nombre1 = paradas[i][0]
    km1 = paradas[i][1]
    nombre2 = paradas[i + 1][0]
    km2 = paradas[i + 1][1]

    distancia = km2 - km1

    print(nombre1, "->", nombre2, ":", distancia, "km")

    if distancia > mayor: 
        mayor = distancia
        tramo = nombre1 + "->" + nombre2

print("Tramo más largo:", tramo, "(", mayor, "km)")
```

```python
# Ejercicio 4. ¿Con chamarra o sin chamarra? 
temps = (8, 14, 22, 27, 31, 19, 11)

def clasificar(t): 
    if t < 12: 
        return "Frío"
    elif t <= 24: 
        return "Templado"
    else: 
        return "Calor"

etiquetas = []
contador = {"Frío": 0, "Templado": 0, "Calor": 0}

for t in temps: 
    resultado = clasificar(t)
    etiquetas.append(resultado)
    contador[resultado] += 1

print(etiquetas)
print(contador)

mayor = 0 
predomina = ""

for clima, cantidad in contador.items():
    if cantidad > mayor: 
        mayor = cantidad 
        predomina = clima

print("Predomina:", predomina)
```

```python
# Ejercicio 5. ¿Cuánto dura el show? 
setlist = (("Intro", 95), ("Riff asesino", 245), ("Balada", 302), ("Solo", 128), ("Encore", 409))

def formato(seg): 
    minutos = seg // 60
    segundos = seg % 60 

    if segundos < 10: 
        return f"{minutos}:0{segundos}"
    else: 
        return f"{minutos}:{segundos}"

total = 0 

for cancion, duracion in setlist: 
    print(cancion + ":", formato(duracion))
    total += duracion 
print("Duración total:", formato(total))
```
