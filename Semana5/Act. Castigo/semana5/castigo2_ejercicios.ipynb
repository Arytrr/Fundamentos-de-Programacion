# BLOQUE 1. La rodada.

**Ejercico 1.** Bitácora de la semana. 

```python
dias = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
km = (120, 85, 0, 210, 95, 0, 150)

def resumen(km):
    total = 0
    dias_rodados = 0
    mayor = 0
    indice = 0

    for i in range(len(km)):
        total += km[i]
        if km[i] > 0:
            dias_rodados += 1
        if km[i] > mayor:
            mayor = km[i]
            indice = i

    promedio = round(total / dias_rodados, 1)

    return total, dias_rodados, promedio, indice

total, dias_rodados, promedio, indice = resumen(km)

print("Total:", total, "km")
print("Días con rodada:", dias_rodados)
print("Promedio por día rodado:", promedio, "km")
print("Día más largo:", dias[indice], "(", km[indice], "km )")
```

**Ejercicio 10.** Traductor de mensaje de WhatsApp.

```python
abrev = {"q": "que", "xq": "porque", "tmb": "también", "pq": "porque", "bn": "bien", "tqm": "te quiero mucho"}

def expandir(mensaje):
    mensaje = mensaje.lower()
    palabras = mensaje.split()
    resultado = []
    for palabra in palabras:
        resultado.append(abrev.get(palabra, palabra))
    return " ".join(resultado)

print(expandir("Q onda tmb voy al concierto xq tocan bn"))
print(expandir("tqm Mamá"))
```

**Ejercicio 11.** Rendimiento de la moto.

```python
def rendimiento():
    while True:
        try:
            km = float(input("Kilómetros recorridos: "))
            litros = float(input("Litros cargados: "))
            resultado = round(km / litros, 1)
            return resultado
        except ValueError:
            print("Escribe solo números")

        except ZeroDivisionError:
            print("Los litros no pueden ser 0")

r = rendimiento()
print(f"Rendimiento: {r} km/l")
if r >= 20:
    print("Excelente")
elif r >= 15:
    print("Normal") 
else:
    print("Revisa tu moto")
```

**Ejercicio 14.** Reserva tu asiento. 

```python
sala = [["X", "L", "L"], ["L", "X", "L"], ["X", "X", "L"]]
def reservar(sala):
    try:
        fila = int(input("Fila: "))
        columna = int(input("Columna: "))
        if fila < 0 or columna < 0:
            raise IndexError
        if sala[fila][columna] == "X":
            print("Ocupado, elige otro")
            return False
        else:
            sala[fila][columna] = "X"
            print("¡Reservado!")
            return True

    except ValueError:
        print("Escribe números")
        return False

    except IndexError:
        print("Ese asiento no existe")
        return False

while True:
    if reservar(sala):
        break
for fila in sala:
    for asiento in fila:
        print(asiento, end=" ")
    print()
```

**Ejercicio 15.** Cupones de descuento.

```python
cupones = {"METAL10": 10, "MOTO20": 20, "ROCK15": 15}
def aplicar(total, codigo):
    codigo = codigo.strip().upper()
    try:
        descuento = cupones[codigo]
        total = total - (total * descuento / 100)
        return total
    except KeyError:
        print("Cupón no válido")
        return total

codigos = [" moto20 ", "GRATIS", "metal10"]
for codigo in codigos:
    resultado = aplicar(1200, codigo)
    print(f"'{codigo}' -> ${resultado}")
```

**Ejercicio 16.** ¿Tu contraseña aguanta?

```python
def evaluar(pw):
    fallas = []
    tienenumero = False
    tienemayuscula = False
    if len(pw) < 8:
        fallas.append("mínimo 8 caracteres")
    for caracter in pw:
        if caracter.isdigit():
            tienenumero = True
        if caracter.isupper():
            tienemayuscula = True
            
    if not tienenumero:
        fallas.append("falta un número")

    if not tienemayuscula:
        fallas.append("falta una mayúscula")

    if " " in pw:
        fallas.append("sin espacios")

    return fallas

contrasenas = ["metal", "Harley2008", "road glide 08", "KLR650moto"]

for pw in contrasenas:
    fallas = evaluar(pw)
    if len(fallas) == 0:
        print(f"{pw}: segura")
    else:
        print(f"{pw}: {fallas}")
```

**Ejercicio 20.** El censor del salón. 

```python
prohibidas = ["tonto", "menso", "baboso"]
def censurar(frase):
    palabras = frase.split()
    contador = 0
    for i in range(len(palabras)):
        if palabras[i].lower() in prohibidas:
            palabras[i] = "*" * len(palabras[i])
            contador += 1

    resultado = " ".join(palabras)
    return resultado, contador

frases = ["no seas menso y pásame el código", "Tonto el que no haga commit", "todo bien profe"]
for frase in frases:
    texto, cantidad = censurar(frase)
    print(f"{texto}  ({cantidad} censuradas)")
```

**Ejercicio 21.** Estacionamiento del foro.

```python
autos = [("ABC123", 9, 11), ("XYZ789", 8, 15), ("MOT650", 10, 11), ("QRO001", 13, 13)]
def cobro(entrada, salida):
    horas = salida - entrada
    if horas <= 1:
        return 20
    else:
        return 20 + (horas - 1) * 15

cobros = {}

for placa, entrada, salida in autos:
    horas = salida - entrada
    precio = cobro(entrada, salida)
    cobros[placa] = precio
    print(f"{placa}: {horas} h -> ${precio}")

total = 0

for precio in cobros.values():
    total += precio
print(f"Total del día: ${total}")
```

**Ejercicio 22.** Tabla del torneo. 

```python
partidos = (("Pumas", "Tigres", 2, 1), ("América", "Pumas", 0, 0), ("Tigres", "América", 3, 2), ("Pumas", "América", 1, 2), ("Tigres", "Pumas", 1, 1))
puntos = {}

for local, visitante, goleslocal, golesvisitante in partidos:
    if local not in puntos:
        puntos[local] = 0

    if visitante not in puntos:
        puntos[visitante] = 0

    if goleslocal > golesvisitante:
        puntos[local] += 3

    elif goleslocal < golesvisitante:
        puntos[visitante] += 3

    else:
        puntos[local] += 1
        puntos[visitante] += 1
print(puntos)

campeon = ""
mayor = 0

for equipo, cantidad in puntos.items():
    if cantidad > mayor:
        mayor = cantidad
        campeon = equipo
print(f"Campeón: {campeon} con {mayor} puntos")
```

**Ejercicio 23.** Pulseras cashles del festival. 

```python
saldos = {"Ana": 500, "Luis": 200, "Mía": 300}
movs = ["Ana:-150", "Luis:-250", "Mía:+100", "Pedro:-50", "Ana:-cien", "Mía:-380", "Luis:-200"]
errores = 0

for movimiento in movs:
    nombre, monto = movimiento.split(":")
    try:
        cantidad = int(monto)
        nuevosaldo = saldos[nombre] + cantidad
        if nuevosaldo < 0:
            print(f"{nombre}: saldo insuficiente")
        else:
            saldos[nombre] = nuevosaldo

    except ValueError:
        print(f"Monto inválido en '{movimiento}'")
        errores += 1

    except KeyError:
        print(f"{nombre} no tiene pulsera")
        errores += 1
print(saldos)
print("Movimientos con error:", errores)
```

**Ejercicio 25.** La playlist del viaje.

```python
playlist = ["Master of Puppets - Metallica - 8:35", "Holy Wars - Megadeth - 6:32",
            "Fade to Black - Metallica - 6:57", "The Trooper - Iron Maiden - 4:10",
            "Hangar 18 - Megadeth - 5:14", "Aces High - Iron Maiden - 4:31"]
def a_segundos(tiempo):
    minutos, segundos = tiempo.split(":")
    return int(minutos) * 60 + int(segundos)

def formato(seg):
    minutos = seg // 60
    segundos = seg % 60
    return f"{minutos}:{segundos:02d}"

artistas = {}
total = 0

for cancion in playlist:
    titulo, artista, tiempo = cancion.split(" - ")
    segundos = a_segundos(tiempo)
    total += segundos
    if artista in artistas:
        artistas[artista] += segundos
    else:
        artistas[artista] = segundos

mayor = 0
mejorartista = ""

for artista, segundos in artistas.items():
    print(f"{artista}: {formato(segundos)}")

    if segundos > mayor:
        mayor = segundos
        mejorartista = artista
print("Más tiempo:", mejorartista)
print("Total:", formato(total))
```
