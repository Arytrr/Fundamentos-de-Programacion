En esta actividad se realizó un programa con un menú que contiene cuatro opciones: tuplas, diccionarios, excepciones y strings. Cada opción permite realizar diferentes operaciones utilizando funciones y los temas aprendidos durante la semana. El programa continúa mostrando el menú hasta que se selecciona la opción de salir. 

```python
# Funciones: "Tuplas".

def sumar_numeros(numeros):
    total = 0
    for numero in numeros:
        total = total + numero
    return total

def tuplas():
    numeros = (1, 7, 11, 13, 27)
    print("Tercer elemento:", numeros[2])

    numero1 = float(input("Ingresa un número: "))
    numero2 = float(input("Ingresa otro número: "))

    n_tupla = numeros + (numero1, numero2)
    print("Nueva tupla:", n_tupla)

    lista = list(n_tupla)
    lista.sort()
    print("Lista ordenada:", lista)

    resultado = sumar_numeros(n_tupla)
    print("Suma total:", resultado)


# Funciones: "Diccionarios".

def buscar_contacto(contactos, nombre):
    if nombre in contactos:
        telefono = contactos[nombre]
        return telefono
    else:
        return None

def diccionarios():
    contactos = { "Fer": "446 110", "Nico": "442 447", "Fany": "5561 8836"}
    nuevo_nom = input("Ingresa el nombre del nuevo contacto: ")
    nuevo_tel = input("Ingresa su teléfono: ")

    contactos[nuevo_nom] = nuevo_tel
    print("Lista de contactos:")

    for nombre in contactos:
        print(nombre)

    buscar = input("Ingresa el nombre que quieres buscar: ")
    telefono = buscar_contacto(contactos, buscar)

    if telefono != None:
        print("Teléfono:", telefono)
    else:
        print(buscar, "no está en tus contactos")


# Funciones: "Excepciones".

def excepciones():
    try:
        numero1 = int(input("Ingresa el primer número entero: "))
        numero2 = int(input("Ingresa el segundo número entero: "))

        suma = numero1 + numero2
        print("Suma:", suma)

        division = numero1 / numero2
        print("División:", division)

    except ValueError:
        print("Error: Debes ingresar dos números enteros.")

    except ZeroDivisionError:
        print("Error: No puedes dividir entre cero.")


# Funciones: "Strings".

def contar_palabras(texto):
    palabras = texto.split()
    cantidad = len(palabras)

    return cantidad

def strings():
    mensaje = "Fuga pa' Medellín, sisas"

    print("Longitud del mensaje:", len(mensaje))
    print("Mensaje en mayúsculas:", mensaje.upper())

    nuevo_men = mensaje.replace("Medellín", "Cali")
    print("Mensaje modificado:", nuevo_men)

    resultado = contar_palabras(mensaje) 
    print("Cantidad de palabras:", resultado)


# Funciones: "Menú Principal". 
opcion = ""
while opcion != "5":

    print("\nMENÚ PRINCIPAL")
    print("1. Tuplas")
    print("2. Diccionarios")
    print("3. Excepciones")
    print("4. Strings")
    print("5. Salir")

    opcion = input("Selecciona una opción: ")
    if opcion == "1":
        tuplas()
    elif opcion == "2":
        diccionarios()
    elif opcion == "3":
        excepciones()
    elif opcion == "4":
        strings()
    elif opcion == "5":
        print("Programa finalizado. ¡Gracias!")
    else:
        print("Opción inválida")
```
