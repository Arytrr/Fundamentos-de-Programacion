```python
# El ticket del merch. 
precios = {"playera": 350, "gorra": 220, "parche": 80, "sudadera": 650}
carrito = {"playera": 2, "parche": 3}
total = 0

for producto, cantidad in carrito.items():
    total += precios[producto] * cantidad

print(f"Total: ${total}")
```
