# Este programa sirve para cobrar los boletos del Museo de Antropología e Historia. Dependiendo de la edad de cada persona, calcula cuánto tiene que pagar y si le corresponde algún descuento. También revisa que la edad sea válida y, al final, muestra cuántos boletos se vendieron y cuánto dinero se recaudó en total.

```python

print("Taquilla Museo Antropología e Historia")

# Registro de ventas
visitantes = 0
dinero = 0

while True:

    respuesta = input("\nIngresa la edad (o 'fin' para terminar): ")

    if respuesta.lower() == "fin":
        break

    # Validar que la edad sea correcta
    try:
        edad = int(respuesta)
    except ValueError:
        print("Error: Esta edad no es valida.")
        continue

    if edad < 0 or edad > 110:
        print("Error: Esta edad no es valida. Debe estar entre 0 y 110 años.")
        continue

    # Determinar los precios de los boletos
    if edad >= 18:
        costo = 45
    elif edad >= 3:
        costo = 30
    else:
        costo = 0

    # Reiniciar el descuento por cada visitante
    porcentaje = 0

    # Aplicar solamente un descuento
    if edad >= 60:
        porcentaje = 0.12

    elif edad >= 18:
        tipo = input("Visitante (1. Estudiante / 2. Profesor / 3. Ninguno): ")

        if tipo not in ["1", "2", "3"]:
            print("Opción incorrecta. Intenta nuevamente.")
            continue

        if tipo == "1" or tipo == "2":
            porcentaje = 0.10

    # Calcular un único descuento
    rebaja = costo * porcentaje
    pago = costo - rebaja

    # Actualizar registro de ventas
    visitantes += 1
    dinero += pago

    print("\n----- BOLETO REGISTRADO -----")
    print("Precio original: $", costo)
    print("Descuento aplicado:", round(porcentaje * 100), "%")
    print("Cantidad descontada: $", round(rebaja, 2))
    print("Total a pagar: $", round(pago, 2))

# Resumen final y corte de caja
print("\n========== CORTE DE CAJA ==========")
print("Boletos vendidos:", visitantes)
print("Dinero recaudado: $", round(dinero, 2))
print("===================================")
```
