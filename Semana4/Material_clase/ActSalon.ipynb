```python
# El gato
tablero = [["X", "O", "X"],
           [" ", "X", "O"],
           ["O", " ", "X"]]

for fila in tablero:
    print(" | ".join(fila))
    if fila != tablero[2]:
        print("---------")

if tablero[0][0] == "X" and tablero[1][1] == "X" and tablero[2][2] == "X":
    print("¡X gana por la diagonal!")
```

```python

def imprimir_matriz():
    tabla = []

    for fila in range(1, 11):
        fila_actual = []
        for columna in range(1, 11):
            fila_actual.append(fila * columna)
        tabla.append(fila_actual)

    for fila in tabla:
        for valor in fila:
            print(valor, end="\t")
        print()

imprimir_matriz()
```
