Este programa crea una tabla de Pitágotas de 10 x 10 utilizando una matriz y la muestra en la pantalla. También permite consultar el resultado de una multiplicación ingresando dos factores, validando que ambos estén dentro del rango del 1 al 10. 

```python
# Código. Parte A: Costrucción de Matriz. 

tabla = []

for fila in range(1, 11):
    fila_actual = []
    for columna in range(1, 11):
        fila_actual.append(fila * columna)
    tabla.append(fila_actual)
```

```python
# Código. Parte B y C: Las dos funciones.

def imprimir_tabla(tabla):
    for fila in tabla:
        for valor in fila:
            print(valor, end="\t")
        print()


imprimir_tabla(tabla)


def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]
```

```python
# Código. Parte D: Programa principal. 

renglon = int(input("Ingresa el primer factor (1-10): "))

while renglon < 1 or renglon > 10:
    print("Número fuera de rango. Intenta nuevamente.")
    renglon = int(input("Ingresa el primer factor (1-10): "))


columna = int(input("Ingresa el segundo factor (1-10): "))

while columna < 1 or columna > 10:
    print("Número fuera de rango. Intenta nuevamente.")
    columna = int(input("Ingresa el segundo factor (1-10): "))


resultado = consultar_producto(tabla, renglon, columna)

print(f"{renglon} x {columna} = {resultado}")
```
