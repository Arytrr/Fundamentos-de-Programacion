```python

from datetime import datetime

# Catalogo de productos
nombres = [
    "Conchas",
    "Cuernos",
    "Fino",
    "Bisquet",
    "Besos",
    "Piedra",
    "Puercos",
    "Orejas",
    "Polvoron",
    "Trebol",
    "Donas",
    "Manoplas",
    "Chinos",
    "Encuercados",
    "Trenzas",
    "Concha grande",
    "Tacos de piña",
    "Cubiletes de crema",
    "Ojos",
    "Bolas rellenas",
    "Tacos de queso con zarzamora",
    "Bollos con queso",
    "Cazuelas",
    "Español",
    "Bolillo",
    "Pan de muerto",
    "Bolsa"
]

precios = [
    10, 10, 10, 10, 10, 10, 10, 10, 10, 10,
    13, 13, 13, 13, 13,
    22, 16, 16, 16, 16, 18, 8, 6, 6, 3, 20, 2
]

# Registro de piezas vendidas
vendidos = [0] * len(nombres)

# Totales del dia
total_diario = 0
total_efectivo = 0
total_tarjeta = 0

continuar = True

while continuar:

    print("\n================================")
    print("     SISTEMA DE LA PANADERIA")
    print("================================")
    print("1. Registrar una venta")
    print("2. Consultar total del dia")
    print("3. Salir")

    try:
        opcion = int(input("Selecciona una opcion: "))

        if opcion == 1:

            total_venta = 0
            productos_venta = []
            respuesta = "SI"

            while respuesta == "SI":

                print("\n========== CATALOGO ==========")
                print("MIGAJON - $10")

                for i in range(10):
                    print(f"{i + 1:02d}. {nombres[i]} - ${precios[i]:.2f}")

                print("\nREPOSTERIA - $13")

                for i in range(10, 15):
                    print(f"{i + 1:02d}. {nombres[i]} - ${precios[i]:.2f}")

                print("\nOTROS PRODUCTOS")

                for i in range(15, len(nombres)):
                    print(f"{i + 1:02d}. {nombres[i]} - ${precios[i]:.2f}")

                try:
                    codigo = int(input("\nIngresa el codigo del producto: "))

                    if codigo >= 1 and codigo <= len(nombres):

                        posicion = codigo - 1

                        print("Producto:", nombres[posicion])
                        print("Precio: $", precios[posicion])

                        cantidad = int(input("Ingresa la cantidad: "))

                        if cantidad > 0:

                            importe = precios[posicion] * cantidad
                            total_venta = total_venta + importe

                            vendidos[posicion] = vendidos[posicion] + cantidad

                            productos_venta.append(
                                [posicion, cantidad, importe]
                            )

                            print("Importe: $", format(importe, ".2f"))

                        else:
                            print("La cantidad debe ser mayor a cero.")

                    else:
                        print("El codigo no existe.")

                except ValueError:
                    print("Debes ingresar un numero valido.")

                respuesta = input(
                    "\nDeseas agregar otro producto? SI o NO: "
                ).upper()

            if total_venta > 0:

                print("\n========== METODO DE PAGO ==========")
                print("1. Efectivo")
                print("2. Tarjeta")

                pago_valido = False

                while not pago_valido:

                    try:
                        pago = int(input("Selecciona el metodo de pago: "))

                        if pago == 1:
                            metodo_pago = "Efectivo"
                            total_efectivo = total_efectivo + total_venta
                            pago_valido = True

                        elif pago == 2:
                            metodo_pago = "Tarjeta"
                            total_tarjeta = total_tarjeta + total_venta
                            pago_valido = True

                        else:
                            print("Selecciona 1 o 2.")

                    except ValueError:
                        print("Ingresa una opcion numerica.")

                total_diario = total_diario + total_venta

                # Generar ticket de compra
                fecha = datetime.now().strftime("%d/%m/%Y %H:%M")

                print("\n================================")
                print("           PANADERIA")
                print("        TICKET DE COMPRA")
                print("================================")
                print("Fecha:", fecha)
                print("--------------------------------")

                for producto in productos_venta:

                    posicion = producto[0]
                    cantidad = producto[1]
                    importe = producto[2]

                    print(nombres[posicion])
                    print(f"{cantidad} x ${precios[posicion]:.2f}")
                    print(f"Importe: ${importe:.2f}")
                    print("--------------------------------")

                print(f"TOTAL: ${total_venta:.2f}")
                print("Metodo de pago:", metodo_pago)
                print("================================")
                print("     GRACIAS POR SU COMPRA")
                print("================================")

            else:
                print("\nNo se registro ningun producto.")

        elif opcion == 2:

            print("\n======= CORTE DE CAJA DEL DIA =======")
            print(f"Total vendido: ${total_diario:.2f}")
            print(f"Total en efectivo: ${total_efectivo:.2f}")
            print(f"Total en tarjeta: ${total_tarjeta:.2f}")

            print("\n===== PIEZAS VENDIDAS POR PRODUCTO =====")

            for i in range(len(nombres)):

                if vendidos[i] > 0:
                    print(nombres[i], ":", vendidos[i], "piezas")

            print("======================================")

        elif opcion == 3:

            continuar = False
            print("Gracias por utilizar el sistema.")

        else:
            print("Opcion no valida. Intenta nuevamente.")

    except ValueError:
        print("Debes ingresar una opcion numerica.")
```
